# Broadcasting 

<p> NumPy broadcasting allows arithmetic operations on arrays with different shapes and sizes by stretching or replicating the smaller array to match the larger one. This enables element-wise operations without explicit looping, making code more concise and efficient. </p>

<h2> How it works: </h2>
<ol>
    <li><b>Dimension Comparison:</b> NumPy compares the shapes of the arrays element-wise from right to left.</li>
    <li><b>Compatibility Rules:</b> Two dimensions are compatible if they are equal or one of them is 1. </li>
    <li><b>Dimension Padding</b>: If the arrays have a different number of dimensions, the smaller array is padded with ones on the left side to match the number of dimensions of the larger array. </li>
    <li><b>Stretching:</b> If the dimensions are compatible, the smaller array is stretched or broadcasted along the larger array's dimensions to match the larger array's shape. </li>
</ol>

```python
import numpy as np
```

```python
# Case 1:

a1 = np.arange(8).reshape(2,4)
a2 = np.arange(8,16).reshape(2,4)
```

```python
print(a1)
print(a2)
```

```python
a1 + a2
```
