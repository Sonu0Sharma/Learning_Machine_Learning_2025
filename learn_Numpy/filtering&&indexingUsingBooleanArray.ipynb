```python
import numpy as np
```

```python
arr1 = np.random.randint(low=1,high=100,size=25).reshape(5,5)
```

```python
arr1
```

# Boolean array==> i.e condition on array

```python
arr1 > 50
```

```python
# odd elements
arr1%2!=0
```

# Indexing using boolean array

```python
arr1[arr1 > 50]
```

```python
# odd elements of arr1
arr1[arr1%2!=0]
```

```python
# Multiple conditions:
# all odd elements that are greater than 50

arr1[(arr1>50) & (arr1%2 !=0)]
```

# Filtering using boolean array

```python
arr1[arr1 > 50] =0
```

```python
arr1
```

```python
#  all values greater than 50 are 0 now
```

```python
#  similarly for odd values
```

```python
arr1[arr1%2!=0]=0
```

```python
arr1
```

```python
arr1[(arr1>50) & (arr1%2 !=0)] =0
```

```python
arr1
```
