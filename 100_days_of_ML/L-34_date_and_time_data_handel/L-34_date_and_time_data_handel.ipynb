# Working with date

```python
#  Conversion

date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

# now we can perform operations -- lets do one my one

## 1. extract year

```python
# make a new col and store data there

date['date']
```

```python
date['date'].dt.year
```

```python
date['date_year'] = date['date'].dt.year
```

```python
date.head()
```

## 2. extarct month

```python
date['date_month_no'] = date['date'].dt.month

date.head()
```

```python
# months by name :

date['date_month_name'] = date['date'].dt.month_name()

date.head()
```

## 3. extract day

```python
date['date_day'] = date['date'].dt.day

date.head()
```

```python
# Date of week

date['date_day_no_of_week'] = date['date'].dt.dayofweek

date.head()
```

```python
# dat name of week

date['date_day_of_week'] = date['date'].dt.day_name()

date.head()
```

```python
# is that day weekday or weekend

date['date_weekend'] = np.where(date['date_day_of_week'].isin(['Saturday','Sunday']),1,0)

date = date.drop(columns=['product_id','city_id','orders'])
```

```python
date.head()
```

## 4. extract week of the year

```python
date['date_week'] = date['date'].dt.isocalendar().week

date.head()
```

## 5. extract quater

```python
date['date_quarter'] = date['date'].dt.quarter
date.head()
```

## 6. extract semester

```python
date['semester'] = np.where(date['date_quarter'].isin([1,2]), 1, 2)

date.head()
```

## 7. Extract Time elapsed between dates

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - date['date']
```

```python
#  lets just seperate days part

(today - date['date']).dt.days
```

## 8. Months passed

```python
# Months passed

today = pd.to_datetime('today')
date['date_diff_months'] = (today.year - date['date'].dt.year) * 12 + (today.month - date['date'].dt.month)

date.head()
```
