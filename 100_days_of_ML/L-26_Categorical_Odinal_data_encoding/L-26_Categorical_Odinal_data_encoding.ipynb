```python
# dependencies

import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('customer.csv')
```

```python
df.sample(5)
```

```python
# we can see 

# categorical col --> gender , review, edication, purchansed

# odinal --> review, education
# so we need to do is :

# gender - nominal data--> one hot encoding
# education, review -- odinal data -- odinal encoding
# purchased --> target -- label encoding
```

```python
# for now

df2 = df.iloc[ : ,2: ]
```

```python
df2.head()
```

```python
#  so as usual 1. train-test-split -> fit on train -> transform both train and test
```

```python
df2.drop('purchased',axis=1)
```

```python
df2['purchased']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(df2.drop('purchased', axis=1),
                                                   df2['purchased'],
                                                   test_size=0.2,
                                                   random_state=1)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
x_test.head()
```

```python
x_test.value_counts()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
# making object of this class , we need to pass a list of lists as parameter, defining in which order we need
# ranking cause otherwise the class will asign the order itself , which may be wrong like in order of 
# poor < good < average

# the list of lists will contain order or each col we are transforming-

oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
```

```python
oe.fit(x_train)
```

```python
# now lets transform

x_train = oe.transform(x_train)
x_test = oe.transform(x_test)
```

```python
x_test
```

# and that's it , ordinal encoding is done ✅

```python
# now lets apply label encoding to the target , -- always remember , traget col is encoded with label encoding
# we dont decide order here

from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
le.fit(y_train)
```

```python
y_train = le.transform(y_train)
y_test = le.transform(y_test)
```
