## 14. Convertors

```python
pd.read_csv('IPL Matches 2008-2020.csv').head()
```

```python
#  lets make a function to apply logic 

def rename (name):
    if name == "Royal Challengers Bangalore":
        return "RBC"
    else:
        return name 
```

```python
#  now apply converter and logic func

pd.read_csv('IPL Matches 2008-2020.csv', converters={'team1': rename}).head()
```
