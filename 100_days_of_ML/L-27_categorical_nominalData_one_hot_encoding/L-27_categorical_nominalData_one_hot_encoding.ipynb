 # now lets encode col with lots of values / categories -- brands

```python
df['brand'].value_counts()
```

```python
# we will do one thing all the names with count <100 , we will club them as others to reduce no of col created

counts = df['brand'].value_counts()

threshold = 100
```

```python
counts[counts <= threshold]
```

```python
counts[counts <= threshold].index
```

```python
replace_var = counts[counts <= threshold].index
```

```python
replace_var
```

```python
pd.get_dummies(df['brand'].replace(replace_var,'others'))
```

```python
pd.get_dummies(df['brand'].replace(replace_var,'others'), dtype= int)
```

```python
pd.get_dummies(df['brand'].replace(replace_var,'others'), dtype= int).sample(5)
```
