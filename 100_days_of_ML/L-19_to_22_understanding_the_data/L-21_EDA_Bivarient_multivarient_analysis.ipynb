```python
# dependencies

import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# data

tips = sns.load_dataset('tips')

titanic = sns.load_dataset('titanic')

flights = sns.load_dataset('flights')

iris = sns.load_dataset('iris')
```

# Now we will start with the multivarient analysis

## Scatter plot --> for numerical & numerical data

```python
# scatter plot on tips data

sns.scatterplot(data= tips, x='total_bill', y='tip')
```

```python
# here we see there is a -- linear relationship in this data

# now lets go for a multi varient analysis , we will check for the gender also , who gives more tip

sns.scatterplot(data= tips, x='total_bill', y='tip', hue='sex')
```

```python
# now we will also add a col for -- smokers

sns.scatterplot(data= tips, x='total_bill', y='tip',
                hue='smoker',style='sex',
                palette='grey' )
```

```python
#  also add another information of the --> size of people

sns.scatterplot(data= tips, x='total_bill', y='tip',
                hue='smoker',style='sex',
                size='size',
                palette='rainbow' )
```

## 3. box plot for (numerical and categorical) col

```python
# lets plot -- average age in each gender 

sns.boxplot(data= titanic, x='sex', y='age')
```

```python
# lets plot -- average age in each gender and also the survival

plt.grid(axis='both', linestyle= '-.')
sns.boxplot(data= titanic, x='sex', y='age', hue='survived')
```

## 4. dist plot for (numerical and categorical) col

```python
# lets find the probability of survival as per age

sns.displot(data=titanic, x='age',hue='survived', kind='kde')
plt.grid(axis='both', linestyle= '-.')
```

# heatmap (categorical + categorical)

```python
# lets find the survived in each pclass

pd.crosstab(titanic['pclass'], titanic['survived'])

# this created the df of the count 
```

```python
sns.heatmap(pd.crosstab(titanic['pclass'], titanic['survived']),
           cmap='Blues')
```

```python
# lets see the exact number of people

sns.heatmap(pd.crosstab(titanic['pclass'], titanic['survived']),
           cmap='Blues',annot=True, fmt='d')
```
