# 2. Numerical Data

histogram

```python
bins = np.arange(0,100,10)

plt.xlabel('Age')
plt.ylabel('Count')

plt.xticks(bins)

plt.grid(axis='both', linestyle='-.')

plt.hist(data['Age'],rwidth= 0.6, bins = bins )
plt.show()
```

```python
# sexy
```

```python
#  now we will plot the dist plot with kde --> y axis shows the probability of a decision

sns.displot(data= data,x=data['Age'])
plt.grid(axis='both', linestyle='-.')
```

```python
sns.displot(data= data,x=data['Age'], kind='kde',)
plt.grid(axis='both', linestyle='-.')
```

```python
# so now we can see here that eg: probability of a random person having age 60 is : 0.005x100 = 0.5%

# kde also tells us about the skewness of data
```

## now lets start with more advance plot- PDF, Boxplot

```python
# Box plot:
```

```python
data.head(2)
```

```python
sns.boxplot(data['Age'])

plt.yticks(np.arange(0,100,5))

plt.grid(axis='both', linestyle='-.')
```

```python
sns.boxplot(data['Age'],orient='h')

plt.xticks(np.arange(0,100,5))

plt.grid(axis='both', linestyle='-.')
```

```python
plt.figure(figsize=(12,5))
sns.boxplot(data['Age'],orient='h')

plt.xticks(np.arange(0,100,2))

plt.grid(axis='both', linestyle='-.')
```

```python
# so now we can see that there are no potential outliers on left side , but there are some on right side
# min age as per our calculation- close to 0
# Q1 start at 20
#meadin - 28
# Q3 = 38
# max as per calculation- 64

# but at last there are few potnetial outliers- 64,6,66,70,71,72,74,80
```
