```python
import pickle
import numpy as np
import pandas as pd

# Load the models
# Note: Ensure the file name is 'clf.pkl', not 'dtc.pkl', as per your
# L-29_titanic_without_pipeline.ipynb notebook.
ohe_sex = pickle.load(open('models/ohe_sex.pkl','rb'))
ohe_embarked = pickle.load(open('models/ohe_embarked.pkl','rb'))
clf = pickle.load(open('models/clf.pkl','rb'))

# Assume user input
# Pclass/gender/age/SibSp/Parch/Fare/Embarked
test_input_df = pd.DataFrame({
    'Pclass': [2],
    'Sex': ['male'],
    'Age': [31.0],
    'SibSp': [0],
    'Parch': [0],
    'Fare': [10.5],
    'Embarked': ['S']
})

# Get the one-hot encoded NumPy arrays
test_input_sex_transformed = ohe_sex.transform(test_input_df[['Sex']])
test_input_embarked_transformed = ohe_embarked.transform(test_input_df[['Embarked']])

# Concatenate the transformed data with the numerical columns
test_input_transformed = np.concatenate([
    test_input_df[['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].values,
    test_input_sex_transformed,
    test_input_embarked_transformed
], axis=1)

# Make the prediction
print(clf.predict(test_input_transformed))
```
