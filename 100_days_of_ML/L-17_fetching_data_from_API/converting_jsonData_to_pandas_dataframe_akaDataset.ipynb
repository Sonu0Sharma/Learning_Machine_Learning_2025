# now this is our beautiful looking data set , now we just need to do in for-loop to get all data and make a final dataset

```python
import pandas as pd
import requests

# Initialising the empty dataframe
dataset = pd.DataFrame()

# for loop, we will need to do string formatting to update page value in loop

for i in range(1, 514):
    response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={}'.format(i))

    temp_df = pd.DataFrame(response.json()['results'])[['id','title','release_date','overview','original_language','adult','vote_count','vote_average','popularity']]

    # Corrected line: Use pd.concat and assign the result back to dataset
    dataset = pd.concat([dataset, temp_df], ignore_index=True)


# now for results:
print("Data collection complete. Here's the head of your dataset:")
print(dataset.head())
print(f"Total rows in dataset: {len(dataset)}")
```

```python
for i in range(1, 514):
    response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={}'.format(i))

    # Check the HTTP status code first
    if response.status_code == 200:
        # Only try to access 'results' if the request was successful
        temp_df = pd.DataFrame(response.json()['results'])[['id','title','release_date','overview','original_language','adult','vote_count','vote_average','popularity']]
        dataset = pd.concat([dataset, temp_df], ignore_index=True)
    else:
        print(f"Error fetching data for page {i}. Status Code: {response.status_code}")
        print(f"Response content: {response.text}") # Print the raw response content
        # You might want to break the loop here or handle the error gracefully
        break # Breaking the loop to prevent further errors if the key is consistently missing

# now for results:
print("Data collection complete. Here's the head of your dataset:")
print(dataset.head())
print(f"Total rows in dataset: {len(dataset)}")
```
