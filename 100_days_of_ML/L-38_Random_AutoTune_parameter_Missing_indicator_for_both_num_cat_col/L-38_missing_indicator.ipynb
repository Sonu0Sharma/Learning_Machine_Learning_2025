```python
# dependencies

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.impute import MissingIndicator,SimpleImputer
```

```python
df = pd.read_csv('l-38_train.csv',usecols=['Age','Fare','Survived'])
```

```python
df.head()
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
X_train.head()
```

# for comparison we will see results of 2 techniques --

##  SimpleImputer(mean).      v/s.         MissingIndicator

```python
# SimpleImputer(mean).


df.isnull().mean() * 100
```

```python
si = SimpleImputer(strategy='mean')
```

```python
X_train_trf = si.fit_transform(X_train)
```

```python
# by default simple imputer will add mean values to missing values
```

```python
X_test_trf = si.transform(X_test)
```

```python
X_train_trf
```

```python
X_test_trf
```

```python
# now lets test wil model 1

from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()

clf.fit(X_train_trf,y_train)

y_pred = clf.predict(X_test_trf)

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

## now we will use Missing Indicator 

```python
mi = MissingIndicator()

mi.fit(X_train)
```

```python
mi.features_
```

```python
# therefore -- 0 th col has missing values 
```

```python
X_train_missing = mi.transform(X_train)
```

```python
X_train_missing
```

```python
X_test_missing = mi.transform(X_test)
```

```python
X_test_missing
```

```python
X_train['Age_NA'] = X_train_missing
```

```python
X_train.head()
```

```python
X_test['Age_NA'] = X_test_missing
```

```python
X_test.head()
```

```python
si = SimpleImputer()

X_train_trf2 = si.fit_transform(X_train)
X_test_trf2 = si.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()

clf.fit(X_train_trf2,y_train)

y_pred = clf.predict(X_test_trf2)

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

As we can see that the performance has improved by 2%, even with same model and simple imputer , this is the benefit for Missing indicator 

## now lets apply missing indicator using sklean -- simple imputer == to make code more easy

```python
si = SimpleImputer(add_indicator=True, strategy='mean')
```

```python
X_train = si.fit_transform(X_train)
```

```python
X_test = si.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()

clf.fit(X_train_trf2,y_train)

y_pred = clf.predict(X_test_trf2)

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
