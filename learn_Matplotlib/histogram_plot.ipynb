```python
import matplotlib.pyplot as plt
import numpy as np

# --- Generate Sample Data ---
# For demonstration, we'll create two sets of random data.
# data1 will be used for single histograms.
# data2 will be used alongside data1 for stacked histograms.
data1 = np.random.randn(1000) # Standard normal distribution
data2 = np.random.randn(1000) * 2 + 3 # Another distribution, shifted and scaled

# --- Set up the Plot Figure ---
# We'll create a figure with a 2x2 grid of subplots to display all four histtypes.
plt.figure(figsize=(12, 10)) # Adjust figure size for better visibility

# --- 1. Bar Histogram ('bar') ---
# This is the default and most common type. Each bin is a solid rectangle.
plt.subplot(2, 2, 1) # (rows, columns, plot_number)
plt.hist(data1, bins=30, histtype='bar', color='skyblue', edgecolor='black', alpha=0.8)
plt.title("Histogram with histtype='bar'")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7) # Add a grid for readability

# --- 2. Stacked Bar Histogram ('barstacked') ---
# Useful for showing the combined distribution of multiple datasets,
# where each dataset's contribution is stacked within a single bar.
plt.subplot(2, 2, 2)
# Pass a list of data arrays to 'x' to stack them
plt.hist([data1, data2], bins=30, histtype='barstacked', color=['lightcoral', 'lightgreen'], edgecolor='black', alpha=0.8)
plt.title("Histogram with histtype='barstacked'")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend(['Dataset 1', 'Dataset 2']) # Add a legend for stacked data
plt.grid(axis='y', linestyle='--', alpha=0.7)

# --- 3. Step Histogram ('step') ---
# Draws the histogram using lines for the bin edges, without filling the area.
# Good for overlaying multiple distributions without obscuring each other.
plt.subplot(2, 2, 3)
plt.hist(data1, bins=30, histtype='step', color='purple', linewidth=2, linestyle='-')
plt.title("Histogram with histtype='step'")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)

# --- 4. Stepfilled Histogram ('stepfilled') ---
# Similar to 'step', but the area under the step lines is filled.
# Provides a clear outline and a solid visual of the distribution's shape.
plt.subplot(2, 2, 4)
plt.hist(data1, bins=30, histtype='stepfilled', color='orange', alpha=0.7, edgecolor='black')
plt.title("Histogram with histtype='stepfilled'")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)

# --- Final Adjustments ---
# Automatically adjust subplot parameters for a tight layout.
plt.tight_layout()
# Display the plot
plt.show()
```

# now plotting 2 data plots in single histogram 

```python
marks_50_students1 = np.random.randint(0, 100, (50))
marks_50_students2 = np.random.randint(0, 100, (50))
```

```python
bins = np.arange(0,100,5)

plt.figure(figsize = (8,6))

plt.hist([marks_50_students1 , marks_50_students2], bins = bins, color = ['pink' , 'cyan'], alpha = 0.6)

plt.xticks(np.arange(0,100,5))

plt.xlabel('Marks')
plt.ylabel('Frequency')

plt.title('Marks of Students of 2 Classes')

plt.grid(axis= 'y', linestyle =':')


plt.show()
```
