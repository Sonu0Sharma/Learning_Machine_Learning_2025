# dependencies

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
plt.style.use('dark_background')
```

# data

```python
study_hours = [2,3,4,4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]
```

## plotting different graphs on this data

```python
plt.scatter(study_hours, marks, s= 7)

plt.xlabel ('Study Hours')
plt.ylabel ('Marks')

plt.xticks(np.arange(0,14,1))
plt.yticks(np.arange(0,110,10))

plt.grid(axis='both',linestyle ='-.', alpha=0.5 )
plt.show()
```

```python
#  now a line plot

plt.xlabel ('Study Hours')
plt.ylabel ('Marks')

plt.xticks(np.arange(0,14,1))
plt.yticks(np.arange(0,110,10))

plt.plot(study_hours, marks, color = 'red')
plt.show()
```

```python
# overlay 2 graphs - line and scatter

plt.xlabel ('Study Hours')
plt.ylabel ('Marks')

plt.xticks(np.arange(0,14,1))
plt.yticks(np.arange(0,110,10))

plt.plot(study_hours, marks, 'c-')
plt.plot(study_hours, marks, 'bo')
plt.show()
```

```python
# histo gram

plt.xticks(np.arange(0,110,5))
plt.yticks(np.arange(0,5,1))
plt.xlabel('Marks')

bins = np.arange(0,100,5)


plt.hist(marks, bins=bins, rwidth=0.6)
plt.show()
```

 This creates bins:
[0, 5), [5, 10), [10, 15), ..., [95, 100)
Notice: it excludes the upper edge of 100, so any values ≥100 are not included in any bin.

Your dataset contains:

marks = [..., 99, 100]
→ 100 is being dropped, hence only 14 values get binned.

But even worse:
Matplotlib hist() by default is left-inclusive and right-exclusive, meaning:

55 appears twice in your list — but both fall into the [55, 60) bin → count as 2.
So even if you have 14 values binned, one bin may contain 2 → only 13 visible bars.

```python


plt.xticks(np.arange(0, 110, 5))
plt.yticks(np.arange(0, 5, 1))

plt.xlabel('Marks')

bins = np.arange(0, 105, 5)

plt.hist(marks, bins=bins, rwidth=0.6, color='green', edgecolor='white')
plt.show()
```
