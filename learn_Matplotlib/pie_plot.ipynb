 # dependencies

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd 
```

# data 

```python
classes = ['Physics', 'Chemistry', 'Maths', 'Science', 'SST']
marks = [89, 45, 78, 23, 90]
```

# basic pie plot

```python
plt.pie(marks, labels= classes)
plt.show()
```

## change colors

```python
colours = ['red','blue','cyan','yellow','orange']

plt.pie(marks, labels= classes, colors= colours)
plt.show()
```

## show percentage

```python
plt.pie(marks, labels= classes, colors= colours, autopct='%0.1f%%')
plt.show()
```

## explode values --> pie pop up out of circle 

```python

explode_values = [0.1, 0, 0, 0, 0]

plt.pie(marks, labels = classes, colors=colours, autopct = '%0.1f%%', explode=explode_values)
plt.show()
```

```python
# the values in array decide which value and how much it pop out, eg: 1st and 4th lable 

explode_values = [0.1, 0, 0, 0.2, 0]

plt.pie(marks, labels = classes, colors=colours, autopct = '%0.1f%%', explode=explode_values)
plt.show()
```

## add shadow 

```python
plt.pie(marks, labels = classes, colors=colours, autopct = '%0.1f%%', 
                                            explode=explode_values, 
                                            shadow = True)
plt.show()
```

## increase size of circle

```python
plt.pie(marks, labels = classes, colors=colours, autopct = '%0.1f%%', 
                                            explode=explode_values, 
                                            radius = 1.6)
plt.show()
```

## change text in pie

```python
textprops = {'fontsize':14, 'color':'teal'}

plt.pie(marks, labels = classes, colors=colours, autopct = '%0.1f%%', 
                                            explode=explode_values, 
                                            radius = 1.6, textprops = textprops)
plt.show()
```

## border in pie and border style

```python
plt.pie(marks, labels = classes, colors=colours, autopct = '%0.1f%%', 
                                            explode=explode_values, 
                                            radius = 1.6, textprops = textprops, 
                                            linewidth=3,linestyle='--', edgecolor='black')
plt.show()
```

```python
# pie has a different method --> wedgeprops to handel all these properties

textprops = {'fontsize':14, 'color':'k'}
wedgeprops = {'linewidth':3, 'linestyle':'--', 'edgecolor':'black'}

plt.pie(marks, labels = classes, colors=colours, autopct = '%0.1f%%', 
                                            explode=explode_values, 
                                            radius = 1.6, textprops = textprops, 
                                            wedgeprops = wedgeprops)
plt.show()
```

## add legend --> to add a scale that tell which color represents what

```python
plt.figure(figsize=(10,10))

# explode values var
pop_up= [0.1,0,0,0.2,0]

# making font style var
font = {'fontsize': 14, 'color':'teal'}

# making border style var
border_style = {'linewidth':2, 'linestyle':'-.', 'edgecolor':'black'}

# plotting
plt.pie(marks, labels= classes, radius=0.8, textprops= font, wedgeprops=border_style, explode= pop_up, colors = colours ,autopct='%0.2f%%' )

# adding title
plt.title('Subjects and average score')

# adding legend
plt.legend(loc='upper right')

plt.show()
```
