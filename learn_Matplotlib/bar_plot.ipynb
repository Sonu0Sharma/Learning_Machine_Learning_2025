# dependencies

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

# data

```python
subjects = ['Maths', 'English', 'Science', 'Social Studies', 'Computer']
marks = [89, 90, 45, 78, 99]
```

# Basic bar plot

```python
plt.bar(subjects , marks)
plt.show()
```

# Change color

```python
# single color change for all bar-

plt.bar(subjects , marks, color = 'red')
plt.show()
```

```python
# single color change for each bar -

plt.bar(subjects , marks, color = ['red','green','blue','cyan','orange'])
plt.show()
```

```python
#  or we can do like this :

colors= ['red','green','blue','cyan','orange']
plt.bar(subjects , marks, color = colors)

plt.show()
```

```python
# NOTE: color is the keyword , so we cant diclare color outside , it will break the code 
# rather we can either directly pass the array to color insiide plt.bar
# or we can create a varibale --> array of colors , then pass that var to color
```

## change width

```python
colors= ['red','green','blue','cyan','orange']
plt.bar(subjects , marks, color = colors , width = 0.5)

plt.show()
```

## change border of bars -> edgecolor

```python
colors= ['red','green','blue','cyan','orange']
plt.bar(subjects , marks, color = colors , width = 0.6, edgecolor = 'black')

plt.show()
```

## change width of this border -> linewidth

```python
colors= ['red','green','blue','cyan','orange']
plt.bar(subjects , marks, color = colors , width = 0.5, edgecolor = 'black', linewidth=4)

plt.show()
```

## change barder style - linestyle

```python
colors= ['red','green','blue','cyan','orange']
plt.bar(subjects , marks, color = colors , width = 0.5, edgecolor = 'black', linewidth= 2, linestyle = '-.')

plt.show()
```

## change the direction of bars : verticals --> horizontal -> barh

```python
plt.barh(subjects , marks, color = colors , edgecolor = 'black', linewidth= 2)

plt.show()
```

# 2 bar plots in same graph

## data

```python
subjects = ['Maths', 'English', 'Science', 'Social Studies', 'Computer']
marks1 = [89, 90, 45, 78, 99]
marks2 = [78, 56, 34, 90, 12]
```

```python
# now graph

plt.figure(figsize=(6,6))

plt.bar(subjects, marks1)
plt.bar(subjects, marks2)

plt.xlabel('Subjects')
plt.ylabel('Marks')

plt.legend('data_')

plt.show()
```

## this was the method we used in line and scatter plots, but here in bar there is one issue if values1 is bigger than value 2 then value2 will be completely hidden

```python
#  so to solve this issue we have a different method here :-


# data 
subjects = ['Maths', 'English', 'Science', 'Social Studies', 'Computer']
marks1 = [89, 90, 45, 78, 99]
marks2 = [78, 56, 34, 90, 12]


width = 0.4

# graph:
plt.figure(figsize = (8,6))

plt.bar(subjects, marks1, width = width)
plt.bar(subjects + width, marks2, width = width)

plt.xlabel('Subjects')
plt.ylabel('Marks')

plt.show()
```

```python
# since subjects here is a list of string and width is a float --> they cant be added so:

# data 
subjects = ['Maths', 'English', 'Science', 'Social Studies', 'Computer']
marks1 = [89, 90, 45, 78, 99]
marks2 = [78, 56, 34, 90, 12]

# changing subjects to a list of int
subjects_len = np.arange(len(subjects))


width = 0.4

# graph:
plt.figure(figsize = (8,6))

plt.bar(subjects_len, marks1, width = width)
plt.bar(subjects_len + width, marks2, width = width)

plt.xlabel('Subjects')
plt.ylabel('Marks')

plt.show()
```

```python
# we can also create great visuals by giving colors and give one group aplha value :

plt.figure(figsize = (8,6))

plt.bar(subjects_len, marks1, width = width, color = colors)
plt.bar(subjects_len + width, marks2, width = width, color = colors, alpha = 0.4)

plt.xlabel('Subjects')
plt.ylabel('Marks')

plt.show()
```

```python
# now to put back the subjects name and also label the group we can do :

import matplotlib.pyplot as plt
import numpy as np

# Data
subjects = ['Maths', 'English', 'Science', 'Social Studies', 'Computer']
marks1 = [89, 90, 45, 78, 99]
marks2 = [78, 56, 34, 90, 12]

# Numeric x positions
x = np.arange(len(subjects))  # [0, 1, 2, 3, 4]
width = 0.4

# Plot bars
plt.figure(figsize=(8, 6))
plt.bar(x - width/2, marks1, width=width, label='Student 1', color = colors)
plt.bar(x + width/2, marks2, width=width, label='Student 2', color = colors, alpha = 0.4)

# Add x-axis labels
plt.xticks(x, subjects)

# Labels
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.title('Marks Comparison')
plt.legend()

plt.show()
```

## for group of : (this is better method ensuring equal spacing and centered groups)

2 bars: shift them ± width/2

3 bars: shift them -width, 0, +width

4 bars: shift them -1.5*width, -0.5*width, +0.5*width, +1.5*width
