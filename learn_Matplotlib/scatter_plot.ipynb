## Adding lables to x,y

### xlabel('') , ylabel('')

```python
# data :

temperature_pune = [25,34,21,45,28,6,43,18,7,2]
humidity_pune = [28, 25,29,20, 26, 50, 19, 29, 52, 55]

temperature_bangalore = [34,35,36,37,28,27,26,25,31,20]
humidity_bangalore = [40, 38, 36, 35, 42, 44, 41, 40, 34, 45]
```

```python
plt.figure(figsize=(8,4))
plt.scatter(temperature_pune , humidity_pune , color='red' )
plt.show()
```

```python
# but we can only apply markersize in plot function
```

```python
plt.figure(figsize=(8,4))

plt.xlabel('temperature_pune')
plt.ylabel('humidity_pune')

plt.plot(temperature_pune , humidity_pune , 'ro' , markersize = 10)
plt.show()
```

## now chaging the intervals on x,y scale :

### xticks (np.arange(0,60,5)) same for yticks --> here the np.arange(lower limit , upper limit , interval size)

```python
plt.figure(figsize=(8,4))

plt.xlabel('temperature_pune')
plt.ylabel('humidity_pune')

plt.xticks(np.arange(0,60,5))
plt.yticks(np.arange(0,60,5))

plt.plot(temperature_pune , humidity_pune , 'co' , markersize = 10)
plt.show()
```

## Multiple Plots on Same Figure

```python
plt.figure(figsize=(8,4))

plt.xlabel('temperature')
plt.ylabel('humidity')

plt.xticks(np.arange(0,60,5))
plt.yticks(np.arange(0,60,5))

plt.plot(temperature_pune , humidity_pune , 'ro' , markersize = 10)
plt.plot(temperature_bangalore , humidity_bangalore, 'co' , markersize = 10)

plt.show()
```

 # taking Iris flower dataset

```python
data = pd.read_csv('IRIS.csv')
```

```python
data.sample(5)
```

## dataset col as x,y

```python
plt.figure(figsize=(8,4))

plt.xlabel('length')
plt.ylabel('width')

plt.plot(data['sepal_length'], data['sepal_width'], 'co' , markersize = 10)

plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.xlabel('length')
plt.ylabel('width')

plt.plot(data['petal_length'], data['petal_width'], 'ro' , markersize = 10)

plt.show()
```

# Transparency --> Alpha - 0= none , 1= full

```python
plt.figure(figsize=(8,4))

plt.xlabel('length')
plt.ylabel('width')

plt.plot(data['petal_length'], data['petal_width'], 'ro' , markersize = 10 , alpha = 0.5)

plt.show()
```

```python
# to add line among data ponts

plt.figure(figsize=(8,4))

plt.xlabel('length')
plt.ylabel('width')

plt.plot(data['petal_length'], data['petal_width'], 'ro' , markersize = 10 , alpha = 0.5 ,  linestyle='solid')

plt.show()
```

## now to achieve markersize in scatter plot we can use -->.    s=10

```python
plt.figure(figsize=(8,4))

plt.xlabel('length')
plt.ylabel('width')

plt.scatter(data['petal_length'], data['petal_width'], color='red' , s = 100 , alpha = 0.5)

plt.show()
```

# Now adding scale to specify which color point represents what

```python
# now lets combine all the previous knowledge

import matplotlib.pyplot as plt

# Create figure
plt.figure(figsize=(10, 5))

# Set axis limits
plt.xlim(0, 10)
plt.ylim(0, 5)

# set intervals
plt.xticks(np.arange(0,11,1))
plt.yticks(np.arange(0,6,1))

# Set labels
plt.xlabel('length')
plt.ylabel('width')

# Plot petal data (red)
plt.scatter(data['petal_length'], data['petal_width'], 
            color='red', label='Petal', s=100, alpha=0.25)

# Plot sepal data (green)
plt.scatter(data['sepal_length'], data['sepal_width'], 
            color='green', label='Sepal', s=100, alpha=0.25)

# Add legend to explain colors
plt.legend(title='Data Type')

# Show plot
plt.title('Petal vs Sepal Dimensions')
plt.show()
```

```python
# now lets combine all the previous knowledge

import matplotlib.pyplot as plt

# setting the dark theme
plt.style.use('dark_background')

# Create figure
plt.figure(figsize=(10, 5))

# Set axis limits
plt.xlim(0, 10)
plt.ylim(0, 5)

# set intervals
plt.xticks(np.arange(0,11,1))
plt.yticks(np.arange(0,6,1))

# Set labels
plt.xlabel('length')
plt.ylabel('width')

# Plot petal data (orange)
plt.scatter(data['petal_length'], data['petal_width'], 
            color='orange', label='Petal', s=100, alpha=0.35)

# Plot sepal data (cyan)
plt.scatter(data['sepal_length'], data['sepal_width'], 
            color='cyan', label='Sepal', s=100, alpha=0.35)

# Add legend to explain colors
plt.legend(title='Data Type')

# Show plot
plt.title('Petal vs Sepal Dimensions')
plt.show()
```
