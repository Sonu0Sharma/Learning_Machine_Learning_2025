```python
import pandas as pd 
```

```python
data = pd.read_csv('food.csv')
```

```python
data.head()
```

## pivot table - helps to analyse the data by molding the dataset and presenting it in a new and more intutive way

```python
data.pivot_table(index=['city'],columns=['item'],values='spends',aggfunc='mean')
```

```python
data.pivot_table(index=['city','gender'],columns=['item'],values='spends',aggfunc='mean')
```

```python
data.pivot_table(index=['city','gender'],columns=['item','frequency'],values='spends',aggfunc='mean')
```
