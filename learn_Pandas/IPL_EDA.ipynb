# making a function of this :

```python
def runs_scored(batsman_name):
    mask20 = data['batsman']== batsman_name
    df20 = data[mask20]
    group50 = df20.groupby('bowling_team')
    return group50['total_runs'].sum().sort_values(ascending=False).head(3)

```
