# Q. find top batsman of each season 

```python
# since we have runs in one dataset and season in another , we need to merge 2 datasets
# we will be using match_id as Primary key for merging

data2.merge(data1,left_on='match_id',right_on='id')
```

```python
# to check if merging was successful: find shape of both

newData = data2.merge(data1,left_on='match_id',right_on='id')

print(data1.shape)
print(data2.shape)
print(newData.shape)
```
