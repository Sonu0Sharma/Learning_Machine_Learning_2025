# Filling missing values using fillna() method

```python
data['Age']
```

```python
data.fillna(0)
```

```python
# not a good practice cause it will fill 0 in string col also 

# best practice :

data['Cabin'].fillna("Not Specified")
```

```python
# now if you want to show these change / save --> use inplace=True
```
