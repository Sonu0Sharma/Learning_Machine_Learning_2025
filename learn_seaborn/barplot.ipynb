```python
# dependencies

import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# data

data= sns.load_dataset('titanic')
```

```python
data.head(3
         )
```

# Bar plot 

```python
sns.barplot(data=data, x='class', y='fare')
```

A bar plot represents an estimate of central tendency for a numeric variable with the height of each rectangle and

the line at top provides some indication of the uncertainty around that estimate using error bars.

```python
# now giving hue--> divide data based on category

sns.barplot(data=data, x='class', y='fare',
           hue='sex')
```

```python
# this -- here we can see that females have paid more in each class 🤣
```

```python
sns.barplot(data=data, x='class', y='fare',
           hue='sex', palette='jet_r')
```

```python
# orient --> change orientation

sns.barplot(data=data, y='class', x='fare',
           hue='sex', palette='coolwarm',
           orient='h')
```

```python
# estimator --> the line at top , by befault its around mean , but we can chnage it to max

sns.barplot(data=data, x='class', y='fare',
           hue='sex', palette='rainbow',
           estimator=np.max)
```

```python
# ci --> confidence interval - the line at top

sns.barplot(x = 'class' , y = 'fare' , data = data ,
            hue = 'sex',
            palette = 'inferno',
            ci=100,
            errcolor = '#7289da',
            errwidth = 3)
```

```python
# saturation --> simply the color of graph , 0=min , 1=max

sns.barplot(x = 'class' , y = 'fare' , data = data ,
            hue = 'sex',
            palette = 'inferno',
            saturation = 0.3)
```
