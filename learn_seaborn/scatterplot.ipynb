# Scatter plot

```python
# dependencies

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# data

titanic_data = sns.load_dataset('titanic')
```

```python
titanic_data.head(3)
```

```python
# plot --> scatter plot = 3 parameters , x,y,data
```

```python
sns.scatterplot(data=titanic_data, x='age', y='fare')
```

```python
# as we can see- most people paid between 0-100 , few paid between 100-300 and only 2 outlier who paid a lot
# so se can see that age doesnt affect much the fair and thats logical also
```

```python
# applying 3rd column- for making category - alive via hue

sns.scatterplot(data=titanic_data,
                x='age', y='fare',
               hue='alive')
```

```python
# and we can also see that pople paid more have a higher survival rate
```

```python
# now lets add another categorizaation - via style - pasangeer class

sns.scatterplot(data=titanic_data,
                x='age', y='fare',
               hue='alive',
               style='class')
```

```python
# we can again see that more deaths were in square data points - third class ,
# and more survivals from circle - first class
```

```python
# now change the colors - palette little

plt.figure(figsize=(10,4))

sns.scatterplot(data=titanic_data,
                x='age', y='fare',
               hue='alive',
               style='class',
               palette='inferno')

plt.title('Titanic Data Analysis')
```

```python
# now lets play with transparency - alpha

plt.figure(figsize=(10,4))

sns.scatterplot(data=titanic_data,
                x='age', y='fare',
               hue='alive',
               style='class',
               palette='gist_rainbow',
               alpha=1)

plt.title('Titanic Data Analysis')
```

```python
plt.figure(figsize=(10,4))

sns.scatterplot(data=titanic_data,
                x='age', y='fare',
               hue='alive',
               style='class',
               palette='gist_rainbow',
               alpha=0.4)

plt.title('Titanic Data Analysis')
```

```python
# lets combine 2 graphs on same plot- line,scatter

plt.figure(figsize=(10,4))

sns.scatterplot(data=titanic_data,
                x='age', y='fare',
               hue='alive',
               style='class',
               palette='inferno',
               alpha=0.4)

sns.lineplot(data=titanic_data, x='age', y='fare',color='red')

plt.title('Titanic Data Analysis')
```

```python
# lets combine 2 graphs on same plot- line,scatter

plt.figure(figsize=(10,4))

sns.scatterplot(data=titanic_data,
                x='age', y='fare',
               hue='alive',
               style='class',
               palette='inferno',
               alpha=0.4)

sns.lineplot(data=titanic_data, x='age', y='alive',color='red')

plt.title('Titanic Data Analysis')
```

```python
plt.figure(figsize=(10, 4))

# Define the desired order for the 'alive' legend entries
alive_order = ['yes', 'no']

# Plot the scatter plot with a specified hue_order
sns.scatterplot(data=titanic_data,
                x='age', y='fare',
                hue='alive',
                style='class',
                palette='inferno',
                alpha=0.4,
                hue_order=alive_order) # <--- This is the key change

# Plot the line plot
sns.lineplot(data=titanic_data, x='age', y='alive', color='red')

plt.title('Titanic Data Analysis')
plt.show()
```

```python
plt.figure(figsize=(10, 4))

# Define the desired order for the 'alive' legend entries
alive_order = ['yes', 'no']

# Define a palette with the colors in the desired order
# Here, we'll use a custom palette for clarity
alive_palette = {'yes': 'sandybrown', 'no': 'darkgray'}

# Plot the scatter plot with the custom palette
sns.scatterplot(data=titanic_data,
                x='age', y='fare',
                hue='alive',
                style='class',
                palette=alive_palette, # Use the custom palette
                alpha=0.4,
                hue_order=alive_order) # Keep the hue_order for good measure

# Plot the line plot
sns.lineplot(data=titanic_data, x='age', y='alive', color='red')

plt.title('Titanic Data Analysis')
plt.show()
```

```python
plt.figure(figsize=(10, 4))

# Define the desired order for the 'alive' legend entries
alive_order = ['no', 'yes']

# Define a palette with the colors in the desired order
# Here, we'll use a custom palette for clarity
alive_palette = {'yes': 'sandybrown', 'no': 'darkgray'}

# Plot the scatter plot with the custom palette
sns.scatterplot(data=titanic_data,
                x='age', y='fare',
                hue='alive',
                style='class',
                palette=alive_palette, # Use the custom palette
                alpha=0.4,
                hue_order=alive_order) # Keep the hue_order for good measure

# Plot the line plot
sns.lineplot(data=titanic_data, x='age', y='alive', color='red')

plt.title('Titanic Data Analysis')
plt.show()
```

```python
# Define the desired order for the y-axis labels
# y_axis_order = ['no', 'yes']

# Create the lineplot
sns.lineplot(data=titanic_data, x='age', y='alive')

# Correctly set the y-axis ticks to the desired order
plt.yticks(ticks=[0, 1], labels=['yes', 'no'])

plt.title('Titanic Data Analysis')
plt.show()
```

```python
# Create the lineplot
sns.lineplot(data=titanic_data, x='age', y='alive')

# Correctly set the y-axis ticks and labels
# We want 'no' at the bottom and 'yes' at the top, so we set the labels first
plt.yticks(ticks=[0, 1], labels=['no', 'yes'])

# Now, invert the y-axis to reverse the order
plt.ylim(1.1, -0.1) # Set the y-limits in reverse order

plt.title('Titanic Data Analysis')
plt.show()
```

```python
# Create the lineplot
ax = sns.lineplot(data=titanic_data, x='age', y='alive')

# The most robust way to reverse the y-axis is to invert it directly
ax.invert_yaxis()

plt.title('Titanic Data Analysis')
plt.show()
```
