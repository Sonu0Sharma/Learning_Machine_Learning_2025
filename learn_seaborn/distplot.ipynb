 # dist plot
 
 A displot in Seaborn is a figure-level function designed to show the distribution of a single variable (univariate distribution) or the relationship between two variables where at least one is a distribution (bivariate distribution). It's a highly flexible and unified interface for plotting distributions, consolidating functionalities that were previously spread across various functions like histplot, kdeplot, and ecdfplot.
 
    1. distplot (Deprecated):
distplot was the primary function in older versions of Seaborn (prior to 0.11.0) for creating distribution plots.
It primarily combined a histogram with a Kernel Density Estimate (KDE) plot and, optionally, a rug plot.


    2. displot (Recommended):
displot was introduced in Seaborn version 0.11.0 as a figure-level function for creating various types of distribution plots.
It acts as a unified interface for several axes-level functions like histplot, kdeplot, and ecdfplot.

In your case, you've correctly identified hist_kws as the place to put those specific arguments for the histogram bars. The issue is that displot expects all arguments that are not data, x, y, hue, row, col, kind, rug, rug_kws, log_scale, legend, palette, hue_order, hue_norm, color, col_wrap, row_order, col_order, height, aspect, or facet_kws to be passed as general keyword arguments (**kwargs) that will then be internally distributed.

The hist_kws parameter you're trying to use is an argument of sns.histplot, but sns.displot expects the specific color, edgecolor, linewidth, and alpha parameters to be passed directly to displot, and it then routes them to histplot's internal rendering. It does not accept hist_kws as a top-level argument for customization of the visual aspects of the bars.

```python
sns.displot(data['time_spend_company'], bins = bins ,
            color='red' , edgecolor='blue' , linewidth=3 , alpha= 0.5
#              color='green',
             
            #  hist_kws={'color':'red' , 'edgecolor':'blue' , 'linewidth':3 , 'alpha': 0.5},
            # kde_kws={'color':'orange' , 'linewidth':3}
             )
plt.xticks(bins)
```
