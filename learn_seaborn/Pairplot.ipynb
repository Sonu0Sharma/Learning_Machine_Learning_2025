```python
# dependencies

import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# data

data= sns.load_dataset('iris')
```

```python
data.head()
```

# Pair plot

```python
sns.pairplot(data)
```

```python
# hue - extra category

sns.pairplot(data, hue='species')
```

```python
# how lets only make the diagonal hist and rest scatter

sns.pairplot(data, hue='species',diag_kind='hist' )
```

```python
# best color i have seen so fat cause it shoes clear seperation:

sns.pairplot(data, hue='species',diag_kind='hist' ,palette='rainbow')
```

```python
# now lets make the data more redable, since upper triangular and lower triangular are same ,lets reduce it

sns.pairplot(data, hue='species',diag_kind='hist' ,
             palette='rainbow' , corner=True)
```

```python
# now to make the differentiation more easy - lets change the markers in scatter

# s- square, o- circle, D- diamond , v- triangle

sns.pairplot(data, hue='species',diag_kind='hist' ,
             palette='rainbow' , corner=True ,
            markers=["o", "v", "D"])
```
